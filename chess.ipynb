```python
# STEP 1 - CREATE BOARD

board = [0] * 8

for i in range(len(board)):
    board[i] = ["__"] * 8


# STEP 2 - PRINT BOARD

def print_board(board):

    for i, row in enumerate(board):

        print(8 - i, end=": ")

        for square in row:
            print(square, end=" ")

        print()

    print("   a  b  c  d  e  f  g  h")


print_board(board)


# STEP 3 - MAP COLUMNS

col_map = {
    "a": 0,
    "b": 1,
    "c": 2,
    "d": 3,
    "e": 4,
    "f": 5,
    "g": 6,
    "h": 7
}


# STEP 4 - WHITE PIECE INPUT

valid_white_pieces = ["rook", "pawn"]

while True:

    white_input = input("Enter white piece and position: ")

    splitted_white = white_input.split(" ")

    if len(splitted_white) != 2:
        print("Invalid input")
        continue

    piece = splitted_white[0]
    coordinate = splitted_white[1]

    if piece not in valid_white_pieces:
        print("Invalid white piece")
        continue

    if len(coordinate) != 2:
        print("Invalid coordinate")
        continue

    col = coordinate[0]
    row = coordinate[1]

    if col not in col_map:
        print("Invalid column")
        continue

    if not row.isdigit():
        print("Invalid row")
        continue

    if int(row) < 1 or int(row) > 8:
        print("Row must be between 1 and 8")
        continue

    column = col_map[col]
    roww = 8 - int(row)

    board[roww][column] = piece

    white_piece = piece
    white_row = roww
    white_column = column

    print_board(board)

    break


# STEP 5 - BLACK PIECES INPUT

valid_black_pieces = ["pawn", "rook", "bishop", "knight", "queen", "king"]

counter = 0

while True:

    black_input = input("Enter black piece and position: ")

    if black_input == "done":

        if counter > 0:
            break

        else:
            print("Add at least one black piece first")
            continue

    if counter == 16:
        print("Maximum black pieces reached")
        break

    splitted_black = black_input.split(" ")

    if len(splitted_black) != 2:
        print("Invalid input")
        continue

    piece = splitted_black[0]
    coordinate = splitted_black[1]

    if piece not in valid_black_pieces:
        print("Invalid black piece")
        continue

    if len(coordinate) != 2:
        print("Invalid coordinate")
        continue

    col = coordinate[0]
    row = coordinate[1]

    if col not in col_map:
        print("Invalid column")
        continue

    if not row.isdigit():
        print("Invalid row")
        continue

    if int(row) < 1 or int(row) > 8:
        print("Row must be between 1 and 8")
        continue

    column = col_map[col]
    roww = 8 - int(row)

    board[roww][column] = "b_" + piece

    counter += 1

    print_board(board)


# STEP 6 - CAPTURE LOGIC

capturable_pieces = []


# ONLY ROOK CAPTURE LOGIC

if white_piece == "rook":

    # UP
    current_row = white_row - 1
    current_col = white_column

    while current_row >= 0:

        square = board[current_row][current_col]

        if square != "__":

            if square.startswith("b_"):
                capturable_pieces.append(square)

            break

        current_row -= 1


    # DOWN
    current_row = white_row + 1
    current_col = white_column

    while current_row < 8:

        square = board[current_row][current_col]

        if square != "__":

            if square.startswith("b_"):
                capturable_pieces.append(square)

            break

        current_row += 1


    # LEFT
    current_row = white_row
    current_col = white_column - 1

    while current_col >= 0:

        square = board[current_row][current_col]

        if square != "__":

            if square.startswith("b_"):
                capturable_pieces.append(square)

            break

        current_col -= 1


    # RIGHT
    current_row = white_row
    current_col = white_column + 1

    while current_col < 8:

        square = board[current_row][current_col]

        if square != "__":

            if square.startswith("b_"):
                capturable_pieces.append(square)

            break

        current_col += 1


# STEP 7 - RESULTS

print()


if len(capturable_pieces) > 0:

    print("Capturable black pieces:")

    for piece in capturable_pieces:
        print(piece)

else:
    print("No capturable black pieces")
```
